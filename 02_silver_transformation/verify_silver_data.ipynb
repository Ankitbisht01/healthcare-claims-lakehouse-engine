```python
%sql
SELECT 
    source_system, 
    COUNT(*) AS total_claims, 
    SUM(total_amount) AS total_dollars,
    MAX(service_date) AS most_recent_service
FROM delta.`abfss://landing-zone@lshc.dfs.core.windows.net/delta/silver_unified_claims`
GROUP BY source_system
ORDER BY total_claims DESC;
```

```python
%sql
SELECT 
    unified_claim_id, 
    source_system, 
    COUNT(*) as duplicate_count
FROM delta.`abfss://landing-zone@lshc.dfs.core.windows.net/delta/silver_unified_claims`
GROUP BY unified_claim_id, source_system
HAVING COUNT(*) > 1;
```
