```python
%sql
SELECT 
    service_date,
    source_system,
    total_claim_count,
    ROUND(total_revenue, 2) AS total_revenue_usd,
    ROUND(max_single_claim_amount, 2) AS highest_single_claim
FROM delta.`abfss://landing-zone@lshc.dfs.core.windows.net/delta/gold_financial_summary_daily`
ORDER BY service_date DESC, source_system;
```

```python
%sql
SELECT 
    patient_id,
    lifetime_visits,
    ROUND(lifetime_spend, 2) AS total_lifetime_spend,
    first_visit_date,
    last_visit_date
FROM delta.`abfss://landing-zone@lshc.dfs.core.windows.net/delta/gold_patient_summary`
ORDER BY lifetime_spend DESC
LIMIT 10;
```
